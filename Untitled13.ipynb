```python
# 1
employees = {
"Ravi": 92,
"Anita": 88,
"Kiran": 92,
"Suresh": 85
}
high=max(employees.values())
top=[]
for i,j in employees.items():
  if j==high:
    top.append(i)
print(f'Top Performers Eligible for Bonus: {top[0]},{top[1]} (Score: {high})')
```

```python
# 2
input_str="Buy mobile phone buy phone online"
input_str=input_str.lower()
s=''
for i in input_str:
  if i.isalnum() or i==' ':
    s+=i
words=s.split()
freq={}
for word in words:
  if word in freq:
    freq[word]+=1
  else:
    freq[word]=1
result={}
for word in freq:
  if freq[word]>1:
    result[word]=freq[word]
print(result)
```

```python
# 3
sensor_readings = [3, 4, 7, 8, 10, 12, 5]
res=[]
for i in range(len(sensor_readings)):
  if sensor_readings[i]%2==0:
    res.append((i,sensor_readings[i]))
print('Valid Sensor Readings (Hour, Value):')
print(res)
```

```python
# 4
emails = [
"ravi@gmail.com",
"anita@yahoo.com",
"kiran@gmail.com",
"suresh@gmail.com",
"meena@yahoo.com"
]
g,y=0,0
for i in emails:
  if 'gmail.com' in i:
    g+=1
  else:
    y+=1
total=len(emails)
gmail=(g/total)*100
yahoo=(y/total)*100
print(f'gmail.com: {int(gmail)}%')
print(f'yahoo.com: {int(yahoo)}%')
```

```python
# 5
sales = [1200, 1500, 900, 2200, 1400, 3000]
total=sum(sales)
avg=total/len(sales)
for i in range(len(sales)):
  if sales[i]>(avg+(0.30*avg)):
    print(f'Day {i+1}:{sales[i]}')
```
